```python
# Agent Technical Analysis
from pydantic import BaseModel
from typing import List

class KeyPoints(BaseModel):
    """Data model for key points"""
    main_scenario: str
    counter_trend_scenario: str
    potential_support_resistance: str
    entry_SL_TP: str 

class TechnicalAnalysis(BaseModel):
    """Data model for technical analysis"""
    full_analysis: str
    key_points: str
```

```python
agent = Agent(
    name="Technical Analysis Agent",
    instructions="""
    You are a technical analysis expert.
    You are given a stock price data and you need to provide a technical analysis of the stock.
    """,
    model = "gpt-5-mini",
    output_model=TechnicalAnalysis,
)
type(agent)
```
